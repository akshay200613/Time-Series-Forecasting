## Stationarity check (ADF)

```python
def check_stationarity(series):
    result = adfuller(series)
    print(f'ADF Statistic: {result[0]}')
    print(f'p-value: {result[1]}')
    if result[1] <= 0.05:
        print("Series is Stationary")
    else:
        print("Series is non-Stationary, differencing required")

check_stationarity(ts_data)
```

```python
# Differencing to remove trend/seasonality for ARIMA
ts_diff = ts_data.diff().dropna()
check_stationarity(ts_diff)
```

```python
ts_diff.plot(figsize=(12,6))
plt.title("Differenced Series")
plt.tight_layout()
plt.plot()
```

```python
fig, axes = plt.subplots(1,2, figsize=(14, 4))
plot_acf(ts_diff, ax=axes[0], lags=40)
plot_pacf(ts_diff, ax=axes[1],lags=40)
plt.suptitle("ACF - PACF - After differencing")
plt.tight_layout()
plt.show()
```

## Modeling

```python
train_size = int(len(ts_data) * 0.8)
train_ts, test_ts = ts_data[:train_size], ts_data[train_size:]
train_exog, test_exog = exog[:train_size], exog[train_size:]
```

```python
forecast_steps= len(test_ts)
print(f"Training samples: {len(train_ts)}")
print(f"Testing samples (steps to forecast): {forecast_steps}")
```

# ARIMA 

```python
model_arima = ARIMA(train_ts, order=(1,1,1))
results_arima = model_arima.fit()
forecast_arima = results_arima.forecast(steps=forecast_steps)[:forecast_steps]
```

```python
plt.figure(figsize=(12,4))
plt.plot(test_ts.values, label='Actual')
plt.plot(train_ts.values, label='Train set')
plt.plot(forecast_arima.values, label='Arima Forecast')
plt.title('ARIMA - Actual vs Forecast')
plt.legend()
plt.tight_layout()
plt.show()
```

# SARIMA

```python
model_sarima = SARIMAX(train_ts,order=(1,1,1), seasonal_order=(1,1,1,52))
results_sarima = model_arima.fit()
forecast_sarima = results_sarima.forecast(steps=forecast_steps)[:forecast_steps]
```

```python
plt.figure(figsize=(12,6))
plt.plot(test_ts.values, label='Actual')
plt.plot(forecast_sarima.values, label='SARIMA Forecast')
plt.title('SARIMA - Actual vs Forecast')
plt.legend()
plt.tight_layout()
plt.show()
```

# Exponential Smoothing(Holt-Winters)

```python
# Exponential Smoothing with additive seasonality to handle the periodic surges
model_es = ExponentialSmoothing(train_ts, trend='add', seasonal='add',seasonal_periods=52 )
result_es = model_es.fit()
forecast_es = result_es.forecast(steps=forecast_steps)[:forecast_steps]
```

```python
plt.figure(figsize=(12,6))
plt.plot(test_ts.values, label='Actual')
plt.plot(forecast_es, label='Exp smoothing Forecast')
plt.title("Exp smoothing - Actual vs Forecast")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
def evaluate(actual, pred, name):
    # Secondary check to ensure identical lengths
    mae = mean_absolute_error(actual, pred)
    rmse = np.sqrt(mean_squared_error(actual, pred))
    return {'Model': name, 'MAE': round(mae, 2), 'RMSE': round(rmse, 2)}

results = []
results.append(evaluate(test_ts, forecast_arima, 'ARIMA'))
results.append(evaluate(test_ts, forecast_sarima, 'SARIMA'))
results.append(evaluate(test_ts, forecast_es, 'Exp Smoothing'))

comparison_df = pd.DataFrame(results)
print("\nModel Performance Table:")
print(comparison_df)
```
